17. Mini KRR System

```r
student_id <- readline("Enter Student ID: ")
attendance <- as.numeric(readline( "Enter Attendance: "))
grade <- as.numeric(readline("Enter Grade: "))

student <- function(student_id, attendance, grade) {

  if (attendance < 75 & grade < 75) {
    risk <- "High Risk"
    recommendation <- "Immediate academic intervention"
  } else if (attendance < 85 | grade < 80) {
    risk <- "Moderate Risk"
    recommendation <- "Academic monitoring"
  } else if (attendance >= 85 & grade >= 80) {
    risk <- "Low Risk"
    recommendation <- "Continue regular monitoring"
  }

cat("\nStudent ID:", student_id, "\n")
cat("Attendance:", attendance,"\n")
cat("Grade:", grade, "\n\n")
cat("Risk Level:", risk, "\n\n")
cat("Recommendation:\n")
cat(recommendation, "\n")
}

student(student_id, attendance, grade)
```
