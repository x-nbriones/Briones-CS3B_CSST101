3. Numerical Operators

```r
x <- 5
y <- 3

add <- x + y
sub <- x - y
div <- x / y
mul <- x * y

print (add, sub)
print (add)
print (sub)
print (div)
print (mul)

cat("\n +", add)
cat("\n-", sub)
cat("\n/", div)
cat("\n*", mul)
```

8. Vectors

```r
grades <- c(98, 86, 96, 83, 91)

grades

cat("\nMean:", mean(grades))
cat("\nMax:", max(grades))
cat("\nMin:", min(grades))
cat("\nSum:", sum(grades))
cat("\nLength:", length(grades))
```
